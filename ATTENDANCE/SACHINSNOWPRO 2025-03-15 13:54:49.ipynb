```
# Import python packages
import streamlit as st
import pandas as pd

# We can also use Snowpark for our analyses!
from snowflake.snowpark.context import get_active_session
session = get_active_session()
```

```
session.sql("""
    CREATE OR REPLACE TABLE employee_attendance (
        employee_id INT,
        employee_name STRING,
        department STRING,
        attendance_date DATE,
        status STRING 
    );
""").collect()
```

```
df_attendance = session.sql("""
    SELECT attendance_date, status, COUNT(*) AS count
    FROM employee_attendance
    GROUP BY attendance_date, status
    ORDER BY attendance_date DESC;
""").to_pandas()

# Show the result
print(df_attendance)
```

```
import streamlit as st
import pandas as pd

st.title("📊 Employee Attendance Dashboard")

# Fetch attendance data
df_attendance = session.sql("""
    SELECT attendance_date, status, COUNT(*) AS count
    FROM employee_attendance
    GROUP BY attendance_date, status
    ORDER BY attendance_date DESC;
""").to_pandas()

# Display the table
st.write("### Attendance Summary")
st.dataframe(df_attendance)

# Plot attendance trends
st.bar_chart(df_attendance.set_index("ATTENDANCE_DATE"))
```
